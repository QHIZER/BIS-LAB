```python
import random

# Distance matrix between 4 cities
dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

n = len(dist)
ants = 10
iterations = 100
alpha, beta, rho, Q = 1, 2, 0.5, 100

pheromone = [[1.0] * n for _ in range(n)]
best_tour = None
best_length = float('inf')

for iteration in range(iterations):
    for ant in range(ants):
        tour = [random.randrange(n)]

        while len(tour) < n:
            current = tour[-1]
            unvisited = [c for c in range(n) if c not in tour]

            weights = [
                pheromone[current][c] ** alpha *
                (1 / dist[current][c]) ** beta
                for c in unvisited
            ]

            next_city = random.choices(
                unvisited, weights=weights
            )[0]
            tour.append(next_city)

        length = sum(
            dist[tour[i]][tour[(i + 1) % n]]
            for i in range(n)
        )

        if length < best_length:
            best_length = length
            best_tour = tour[:]

    # Evaporate pheromones
    for i in range(n):
        for j in range(n):
            pheromone[i][j] *= (1 - rho)

    # Deposit pheromones from the best tour
    for i in range(n):
        a = best_tour[i]
        b = best_tour[(i + 1) % n]
        pheromone[a][b] += Q / best_length
        pheromone[b][a] += Q / best_length

# Detailed output
route = best_tour + [best_tour[0]]

print("===== ACO FOR TSP =====")
print("Number of Cities:", n)
print("Number of Ants:", ants)
print("Number of Iterations:", iterations)
print("Best Tour:", " -> ".join(map(str, route)))
print("\nDistance Details:")

total = 0
for i in range(len(route) - 1):
    a, b = route[i], route[i + 1]
    d = dist[a][b]
    total += d
    print(f"City {a} -> City {b}: {d}")

print("\nShortest Total Distance:", total)
print("Starting City:", best_tour[0])
print("All Cities Visited:", len(set(best_tour)) == n)
print("Returns to Origin:", route[0] == route[-1])
```
